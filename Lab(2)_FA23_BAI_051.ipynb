# **Task 3: Addition & Subtraction**


---


•	Create two 3×3 matrices representing the marks of 3 students in 3 subjects for Term 1 and Term 2.

•	Calculate the total combined marks (Term 1 + Term 2) for each student in each subject.

•	Calculate the improvement (Term 2 − Term 1) for each student in each subject.

•	Using the improvement matrix, identify (through code, not by manually reading) which student had the highest total improvement across all subjects.


```python
#Term 1 marks
students = np.array(["Nimra", "Ayesha", "Fatima"])

term1 = np.array([
    [70, 75, 80],
    [65, 80, 72],
    [85, 78, 88]
])

print("Term 1 Marks:")
print(term1)
```

```python
#Term 2 marks
term2 = np.array([
    [78, 82, 85],
    [75, 85, 80],
    [90, 84, 92]
])

print("Term 2 Marks:")
print(term2)
```

```python
#Total combined marks
total_marks = term1 + term2

print("Total Combined Marks:")
print(total_marks)
```

```python
#Improvement
improvement = term2 - term1

print("Improvement Matrix:")
print(improvement)
```

```python
#Student with highest total improvement

total_improvement = np.sum(improvement, axis=1)

print("Total Improvement:")
print(total_improvement)

highest_index = np.argmax(total_improvement)

print("Student with highest total improvement:")
print(students[highest_index])
```

# **Task 4: Dot Product**


---


•	Create a vector of 4 item prices and a vector of the quantities purchased for each item.

•	Calculate the total bill using the dot product (not by multiplying and adding manually).

•	Create a 2×4 matrix representing quantities purchased by 2 different customers, then calculate the total bill for both customers using a single dot product operation.

•	Explain (in a comment in your code) why the shapes of the two arrays had to match a specific way for the dot product to work.


```python
import numpy as np
#Prices and quantities
prices = np.array([10.5, 5.2, 20.0, 1.75])
quantities = np.array([3, 10, 2, 5])

print(f"Item Prices: {prices}")
print(f"Quantities Purchased: {quantities}")
```

```python
#Total bill using dot product
total_bill = np.dot(prices, quantities)
print(f"\nTotal bill for single customer: ${total_bill:.2f}")
```

```python
#Quantities of two customers
quantities_customers = np.array([
    [3, 10, 2, 5],
    [1, 5, 3, 2]
])

print("\nQuantities purchased by two customers:")
print(quantities_customers)

#Bills for both customers
total_bills_customers = np.dot(quantities_customers, prices)
print(f"\nTotal bills for customers: ${total_bills_customers[0]:.2f}, ${total_bills_customers[1]:.2f}")
```

```python
# customer_quantities has shape (2, 4)
# prices has shape (4,)
#
# The 4 columns of customer_quantities must match
# the 4 elements in the prices vector.
#
# Therefore, the dot product produces one total bill
# for each of the 2 customers.
```

•	Extract every alternate row and every alternate column from the full matrix using step slicing.

```python
 #Top-left 3×3
top_left_3x3 = matrix_6x6[0:3, 0:3]
print("\nTop-left 3x3 sub-matrix:")
print(top_left_3x3)
```

```python
#Alternate rows and columns
alternate_rows_cols = matrix_6x6[::2, ::2]
print("\nEvery alternate row and column:")
print(alternate_rows_cols)
```

```python
#Last 2 rows and columns
last_2_rows_cols = matrix_6x6[-2:, -2:]
print("\nLast 2 rows and last 2 columns (sub-matrix):")
print(last_2_rows_cols)
```
